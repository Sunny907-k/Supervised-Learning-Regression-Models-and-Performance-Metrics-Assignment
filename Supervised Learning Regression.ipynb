Question 1 : What is Simple Linear Regression (SLR)? Explain its purpose.

Ans - Simple Linear Regression (SLR) is a statistical method used to model and analyze the relationship between two quantitative variables:

one independent variable (X) — also called the predictor or explanatory variable, and

one dependent variable (Y) — also called the response or outcome variable.

Its purpose is to predict the value of the dependent variable (Y) based on the known value of the independent variable (X).

Question 2: What are the key assumptions of Simple Linear Regression?

Ans - Assumption of Linear Regression -

1) Linearity - X and Y should have linear relationship.
2) Independance - The observations rows are independant to each other.
3) Homoscedasticity - The variance of errors are constant.
4) Normality of Error - Errors should be normally distributed.
5)Multicollinearity - Features should have least relation.

Question 4: Provide a real-world example where simple linear regression can be
applied.

Ans - We can use simple linear regression to predict the price of a house based on its size.

We make a model like this:

House Price = mx + c

House Price (Y): what we want to predict

Size (X): the number of square feet

β₀ (intercept): price when the size is 0 (starting value)

β₁ (slope): how much the price increases when size increases by 1 sq. ft.

Question 7: Name and briefly describe three common evaluation metrics for regression models.

Ans - Three Common Evaluation Metrics for Regression Models (Simple Explanation)

1) Mean Absolute Error (MAE) - It measures the average difference between the predicted values and the actual values.

2) Mean Squared Error (MSE) - It measures the average of the squared differences between predicted and actual values.

3) R-squared (R²) - It shows how well the model fits the data.

Question 8: What is the purpose of the R-squared metric in regression analysis?

Ans - R-squared (R²) shows how well a regression model fits the data.

In simple words:

It tells us how much of the change in the dependent variable (Y) can be explained by the independent variable (X).

The value of R² is between 0 and 1:

R² = 1 → Perfect fit (the model explains all the variation)

R² = 0 → No fit (the model explains nothing)

Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.


```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2, 4, 5, 4, 5])

model = LinearRegression()
model.fit(X, Y)

print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

Question 10: How do you interpret the coefficients in a simple linear regression model?

Ans - In a simple linear regression model, the intercept (β₀ or c) represents the value of the dependent variable (Y) when the independent variable (X) is zero, while the slope (β₁ or m) represents how much Y changes for a one-unit increase in X. For example, if β₀ = 50 and β₁ = 5, it means Y starts at 50 when X is 0, and increases by 5 for every 1-unit increase in X.
